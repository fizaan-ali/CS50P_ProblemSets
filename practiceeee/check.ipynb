```python
from fpdf import FPDF
```

```python
pdf = FPDF( orientation = 'P', format = 'A4' )
```

```python
pdf.add_page()
```

```python
pdf.set_font('Helvetica', style='I', size=16)
```

```python
pdf.cell(15, 15, "Fizaan Ali", border=1, ln=1, align='C')
```

```python
pdf.image('shirtificate.png', x=120 ,y=30, w=100, h=100)
```

```python
pdf.output('first.pdf')
```
